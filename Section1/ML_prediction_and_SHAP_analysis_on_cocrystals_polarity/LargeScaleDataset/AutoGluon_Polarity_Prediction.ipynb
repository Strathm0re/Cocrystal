```python
import pandas as pd
from autogluon.tabular import TabularPredictor
import numpy as np


# Load data
data_path = "./Large_Scale_dataset.csv"
data = pd.read_csv(data_path)
data.drop(columns=['pair_name'], inplace=True)

# Inspect data
print("Data shape:", data.shape)
print("First few rows of the data:")
print(data.head())

# Identify target column (last column)
target_column = data.columns[-1]
print("Target column:", target_column)

# Define evaluation metrics and presets
eval_metric = 'recall_macro'
presets = 'medium'
num_bag_folds = 5

# Initialize result storage
results = []

i = 1
print(f"Training model_{i}...")

model_path = f'./model/model_{i}'
predictor = TabularPredictor(path=model_path, label=target_column, problem_type='binary', eval_metric=eval_metric).fit(
    train_data=data, presets=presets, calibrate_decision_threshold=True, num_bag_folds=num_bag_folds, refit_full='all', auto_stack=True, num_stack_levels=0
)
# Evaluate model
eval_results = predictor.evaluate(data, silent=True)
# Extract evaluation results
accuracy = eval_results['accuracy']
balanced_accuracy = eval_results['balanced_accuracy']
# Store results in list
results.append({
    'Model': f'model_{i}',
    'Accuracy': accuracy,
    'Balanced_Accuracy': balanced_accuracy
})
# Convert results to DataFrame
results_df = pd.DataFrame(results)
# Print results
print("\nResults Table:")
print(results_df)    
```

```python
import pandas as pd
import numpy as np
import shap
import matplotlib.pyplot as plt
from autogluon.tabular import TabularPredictor
import os

# Configure font settings (Arial, 14pt)
plt.rcParams["font.family"] = ["Arial", "sans-serif"]
plt.rcParams["font.size"] = 14
plt.rcParams["axes.unicode_minus"] = False  # Fix negative sign display

# Define file paths
candidate_path = r"F:\project\cocrystal\git_0714\Section1\ML prediction and SHAP analysis on cocrystals polarity\LargeScaleDataset\candidate_structure_combined_features_processed.csv"
large_scale_path = r"F:\project\cocrystal\git_0714\Section1\ML prediction and SHAP analysis on cocrystals polarity\LargeScaleDataset\Large_Scale_dataset.csv"
model_path = "./5best_model/model_1"  # Use the first best model
output_dir = "./shap_plots"
os.makedirs(output_dir, exist_ok=True)

# Load and prepare data
def load_data():
    # Load full candidate dataset
    candidate_data = pd.read_csv(candidate_path)
    
    # Load large scale dataset and take last 51 samples
    large_scale_data = pd.read_csv(large_scale_path)
    large_scale_sample = large_scale_data.tail(51).copy()
    
    # Identify and remove ID columns, keep only features
    id_cols = ['pair_name', 'CA_name', 'candidate_name']
    
    # Process candidate dataset features
    existing_id_cols_candidate = [col for col in id_cols if col in candidate_data.columns]
    candidate_features = candidate_data.drop(columns=existing_id_cols_candidate, errors='ignore')
    
    # Process large scale dataset features
    existing_id_cols_large = [col for col in id_cols if col in large_scale_sample.columns]
    large_scale_features = large_scale_sample.drop(columns=existing_id_cols_large, errors='ignore')
    
    # Ensure feature consistency between datasets
    common_features = candidate_features.columns.intersection(large_scale_features.columns)
    candidate_features = candidate_features[common_features]
    large_scale_features = large_scale_features[common_features]
    
    print(f"Candidate dataset - Samples: {len(candidate_features)}, Features: {len(common_features)}")
    print(f"Large scale dataset - Samples: {len(large_scale_features)}, Features: {len(common_features)}")
    
    return candidate_features, large_scale_features

# Load model and create SHAP explainer
def load_model_and_create_explainer(features, model_path):
    print(f"Loading model: {model_path}")
    predictor = TabularPredictor.load(model_path)

    # Create SHAP explainer (TreeExplainer for tree-based models)
    explainer = shap.Explainer(predictor.predict,features)
    
    # Calculate SHAP values (using sample background data for efficiency)
    background = features.sample(min(100, len(features)), random_state=42)
    shap_values = explainer(features)
    
    # For classification problems, focus on class 1 SHAP values
    if isinstance(shap_values, list) and len(shap_values) > 1:
        class_labels = predictor.class_labels
        if 1 in class_labels:
            class_1_idx = class_labels.index(1)
            shap_values = shap_values[class_1_idx]
        else:
            shap_values = shap_values[0]
    
    return explainer, shap_values, features.columns

# Generate and save SHAP visualizations
def generate_shap_plots(explainer, shap_values, feature_names, dataset_name):
    # Generate beeswarm plot
    plt.figure(figsize=(12, 8))
    shap.plots.beeswarm(shap_values, show=False)
    plt.title(f'{dataset_name} - SHAP Value Beeswarm Plot', fontsize=16)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, f'{dataset_name}_beeswarm.png'), dpi=300, bbox_inches='tight')
    plt.close()
    
    # Generate bar plot
    plt.figure(figsize=(12, 8))
    shap.plots.bar(shap_values, show=False)
    plt.title(f'{dataset_name} - Feature Importance Bar Plot', fontsize=16)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, f'{dataset_name}_bar.png'), dpi=300, bbox_inches='tight')
    plt.close()
    
    print(f'SHAP plots for {dataset_name} saved to {output_dir}')

# Main function
def main():
    # Load data
    candidate_features, large_scale_features = load_data()
    
    # Perform SHAP analysis for both datasets
    for features, name in [
        (candidate_features, 'Candidate_Dataset'),
        (large_scale_features, 'Large_Scale_Dataset_Last51')
    ]:
        explainer, shap_values, feature_names = load_model_and_create_explainer(features, model_path)
        generate_shap_plots(explainer, shap_values, feature_names, name)
    
    print("SHAP analysis completed. All plots have been saved.")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import shap
import numpy as np
from autogluon.tabular import TabularPredictor

# ---------------- 1. 读取 & 清洗 ----------------
# 文件 1：cocrystal 数据
path1 = r"F:\project\cocrystal\git_0714\Section1\ML prediction and SHAP analysis on cocrystals polarity\LargeScaleDataset\candidate_structure_combined_features_processed.csv"
df1 = pd.read_csv(path1).dropna(subset=['is_polar'])
X1 = df1.drop(columns=[df1.columns[-1], 'candidate_name', 'CA_name'])
y1 = df1[df1.columns[-1]]

# 文件 2：Large_Scale_dataset 数据
path2 = "./Large_Scale_dataset.csv"
df2 = pd.read_csv(path2)
X2 = df2.drop(columns=[df2.columns[-1], 'pair_name'])
y2 = df2[df2.columns[-1]]

# ---------------- 2. 拼接 ----------------
X_all = pd.concat([X1, X2], axis=0, ignore_index=True)
y_all = pd.concat([y1, y2], axis=0, ignore_index=True)

# ---------------- 3. SHAP 计算 & 可视化 ----------------
predictor = TabularPredictor.load('./5best_model/model_1')
explainer = shap.Explainer(predictor.predict, X_all)
shap_values = explainer(X_all)
shap.summary_plot(shap_values, X_all)

# ---------------- 4. 打印末 51 条 pair_name ----------------
# 仅来自第二个文件（Large_Scale_dataset.csv）
last_n = 51
pair_names_last = df2['pair_name'].tail(last_n).reset_index(drop=True)
print("末 51 个样本的 pair_name:")
for idx, name in enumerate(pair_names_last, start=len(df2)-last_n+1):
    print(f"{idx}: {name}")
```

```python
import pandas as pd
import shap
import matplotlib.pyplot as plt
import numpy as np
from autogluon.tabular import TabularPredictor

# Load model
model_large_dataset_path = './model/model_1'
predictor = TabularPredictor.load(model_large_dataset_path)

# Load data
data_path = './candidate_structure_combined_features_processed.csv'
data = pd.read_csv(data_path)

# Filter specific samples
candidate_names = ['candidate14_0_1_opt', 'candidate23_+1_1_opt']
specific_samples = data[data['candidate_name'].isin(candidate_names)]

# Ensure data contains only numerical features (remove non-numerical columns)
specific_samples = specific_samples.drop(columns=['candidate_name'])  # Remove non-numerical column
specific_samples = specific_samples.select_dtypes(include=[np.number])  # Ensure all columns are numerical

# Create SHAP Explainer
explainer = shap.Explainer(predictor.predict, specific_samples)

# Calculate SHAP values
shap_values = explainer(specific_samples)


candidate_name = 'candidate14_0_1_opt'
shap.plots.waterfall(shap_values[1], max_display=10, show=False)
plt.title(f'SHAP Waterfall Plot for {candidate_name}')
plt.show()
candidate_name = 'candidate23_+1_1_opt'
shap.plots.waterfall(shap_values[0], max_display=10, show=False)
plt.title(f'SHAP Waterfall Plot for {candidate_name}')
plt.show()
    
```
